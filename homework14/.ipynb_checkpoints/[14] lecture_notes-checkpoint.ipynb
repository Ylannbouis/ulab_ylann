```python
# 1) Units and Quantities
# Help keep calculations consistent
# In 1999, the Mars Climate Orbiter burn up in the Martian atmosphere
# this happened because engineers failed to convert from imperial units
# to metric units causing a miscalulation upon descent
# This all could have been avoided with Astropy!
```

```python
# Defining and Converting units
from astropy import units as u 

# Define a distance of 10 meters, .u function
distance = 10 * u.meter
print(distance)

# Convert meters to centimeters, .to function
print(distance.to(u.centimeter)) 

# How would I convert to kilometers?
print(distance.to(u.kilometer))
```

```python
# Combine Units
speed = 30 * u.kilometer / u.hour

# How would I convert to meters per hour?
print(speed.to(u.meter / u.hour))

# How would I convert to meters per second?
print(speed.to(u.meter / u.second))
```

```python
# Check Unit Consistency
force = 5 * u.newton
mass = 2 * u.kilogram
acceleration = force / mass

# We will get an error if we try to convert to wrong units
# It will catch your mistake! 
# print(acceleration.to(u.gram / u.second **2))
print(acceleration.to(u.meter / u.second **2))
```

```python
# 2) Time and Dates
# Astronomers observe! It's important to observe at the right time
```

```python
# Creating and converting "time objects"
from astropy.time import Time
time1 = Time("2025-02-22 12:00:00")
print(time1)

# Convert to Julian Date (weird thing that astronmers use)
# Historical purposes
print(time1.jd)

# Get the current time (in UTC, its not PST)
now = Time.now()
print(now)

# How would I get "now" in Julian Time?
print(now.jd)
```

```python
# Time Arithmetic
time2 = Time("2025-02-22 16:30:00")
print(time2)

# Calculate the difference in times
delta = time2 - time1
print(delta)

# Convert to hours, .to_value function
hours = delta.to_value('hour')
print(hours)
minutes = delta.to_value('minute')
print(minutes)

# How would I assign units to these values?
print(hours * u.hour)
print(minutes * u.minute)
```
