# 2 Plotting a Function
Goal: Write a **function** that will plot $f(x) = x^2$ and $g(x) = \sqrt{x}$ on the same plot and return the $f(x)$ and $g(x)$ values as separate lists. Your plot needs to have the following:
- An appropriate figure size (`figsize=(10,8)` is recommended)
- Labeled x and y axis
- A title
- Different colors and linestyles for each function
- A legend
- Your domain should be $x \in [0, 10]$. *Hint: Use `np.linspace` to set a range of values for your functions.*

Your function should return:
- A list of $f(x)$ values
- A list of $g(x)$ values.

```python
# Your code here
def plot_new_functions():
    # Plots both x^2 and sqrt(x) on the same graph. Inputs: nothing. Outputs: values of f(x) and g(x) as separate lists.
    plt.figure(figsize=(10,4))
    x_data = np.linspace(0, 10, 10) # Gives a list of size 1000 of numbers from 0 to 10, equally spaced.
    f = x_data**2
    g = np.sqrt(x_data)
    plt.title("plots of x^2 and sqrt(x)")
    plt.xlabel("X-Axis")
    plt.ylabel("Y-Axis")
    plt.plot(x_data, f, label="x^2", linestyle='--', color="blue")
    plt.plot(x_data, g, label="sqrt(x)", linestyle='-', color="red")
    plt.legend()
    plt.show()
    return (f,g)
plot_new_functions()
```

```python
# Import your module and/or your code here
def plot_paraboloid(a,b):
    # Plots a paraboloid in the form of ax^2 + by^2 = z. Inputs: a, b. Outputs: graph of the function
    fig, ax = plt.subplots(subplot_kw={"projection": "3d"})

    X = np.linspace(-1, 1, 100)
    Y = np.linspace(-1, 1, 100)
    X, Y = np.meshgrid(X, Y) # Basically creates a gird of points (X, Y)
    Z = a * (X ** 2) + b * (Y ** 2)

    ax.plot_surface(X, Y, Z, label="Paraboloid", linewidth=0, antialiased=False)
    ax.set_zlim(0, 2)
    ax.set_xlabel("X-Axis")
    ax.set_ylabel("Y-Axis")
    ax.set_zlabel("Z-Axis")
    ax.set_title("Paraboloid")
    plt.legend()

    plt.show()
plot_paraboloid(1,1)
    
```
